```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

display(Markdown("### 4. Convolution of $x[n] = (1/3)^n(u[n]-u[n-51])$ and $y[n] = (1/5)^n u[n]$"))

n_base4 = np.arange(-2, 15)
n = n_base4
x4 = ((1.0 / 3.0) ** n) * ((n >= 0).astype(float) - (n >= 51).astype(float))
y4 = ((1.0 / 5.0) ** n) * (n >= 0).astype(float)

z4 = np.zeros_like(n, dtype=float)
for idx, val in enumerate(n):
    if val < 0:
        z4[idx] = 0.0
    elif 0 <= val <= 50:
        z4[idx] = 1.5 * ((1.0 / 5.0) ** val) * (((5.0 / 3.0) ** (val + 1)) - 1.0)
    else:
        z4[idx] = 1.5 * ((1.0 / 5.0) ** val) * (((5.0 / 3.0) ** 51) - 1.0)

fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
axes[0].stem(n, x4, linefmt='b-', markerfmt='bo', basefmt='k-')
axes[0].set_title(r'Signal $x[n] = (1/3)^n(u[n]-u[n-51])$', fontsize=9.5, fontweight='bold', color='darkblue')
axes[0].set_ylabel('Amplitude', fontsize=8.5)
axes[0].grid(True, linestyle='--', alpha=0.6)

axes[1].stem(n, y4, linefmt='g-', markerfmt='go', basefmt='k-')
axes[1].set_title(r'Signal $y[n] = (1/5)^n u[n]$', fontsize=9.5, fontweight='bold', color='darkgreen')
axes[1].set_ylabel('Amplitude', fontsize=8.5)
axes[1].grid(True, linestyle='--', alpha=0.6)

axes[2].stem(n, z4, linefmt='r-', markerfmt='ro', basefmt='k-')
axes[2].set_title(r'Convolution Result $z[n] = x[n] * y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
axes[2].set_xlabel('Index $n$', fontsize=8.5)
axes[2].set_ylabel('Amplitude', fontsize=8.5)
axes[2].grid(True, linestyle='--', alpha=0.6)
plt.show()
```
