```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

display(Markdown("### 2. Convolution of $x[n] = (1/2)^{n-4}u[n]$ and $y[n] = (1/3)^n u[n-2]$"))

n_base2 = np.arange(-2, 12)
n = n_base2
u1 = (n >= 0).astype(float)
u2 = (n - 2 >= 0).astype(float)
x2 = ((0.5) ** (n - 4)) * u1
y2 = ((1.0 / 3.0) ** n) * u2

z2 = np.zeros_like(n, dtype=float)
for idx, val in enumerate(n):
    if val >= 2:
        z2[idx] = 32.0 * ((1.0 / 3.0) ** val) * (((3.0 / 2.0) ** (val - 1)) - 1.0)
    else:
        z2[idx] = 0.0

fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
axes[0].stem(n, x2, linefmt='b-', markerfmt='bo', basefmt='k-')
axes[0].set_title(r'Signal $x[n] = (1/2)^{n-4}u[n]$', fontsize=9.5, fontweight='bold', color='darkblue')
axes[0].set_ylabel('Amplitude', fontsize=8.5)
axes[0].grid(True, linestyle='--', alpha=0.6)

axes[1].stem(n, y2, linefmt='g-', markerfmt='go', basefmt='k-')
axes[1].set_title(r'Signal $y[n] = (1/3)^n u[n-2]$', fontsize=9.5, fontweight='bold', color='darkgreen')
axes[1].set_ylabel('Amplitude', fontsize=8.5)
axes[1].grid(True, linestyle='--', alpha=0.6)

axes[2].stem(n, z2, linefmt='r-', markerfmt='ro', basefmt='k-')
axes[2].set_title(r'Convolution Result $z[n] = x[n] * y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
axes[2].set_xlabel('Index $n$', fontsize=8.5)
axes[2].set_ylabel('Amplitude', fontsize=8.5)
axes[2].grid(True, linestyle='--', alpha=0.6)
plt.show()
```
