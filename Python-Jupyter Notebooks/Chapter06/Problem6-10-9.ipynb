```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

display(Markdown("### 6. Convolution of Finite Sequence Signals (Proakis & Manolakis 2.21)"))

# x[n] = {1, 2, 1, 1} with origin at index 0 corresponding to value 1 (n = -2 to 1)
# y[n] = {1, -1, 0, 0, 1, 1} with origin at index 0 corresponding to value 1 (n = 0 to 5)
n_x = np.arange(-2, 2)
x6 = np.array([1.0, 2.0, 1.0, 1.0])

n_y = np.arange(0, 6)
y6 = np.array([1.0, -1.0, 0.0, 0.0, 1.0, 1.0])

z6 = np.convolve(x6, y6)
n_z6 = np.arange(n_x[0] + n_y[0], n_x[0] + n_y[0] + len(z6))

fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
axes[0].stem(n_x, x6, linefmt='b-', markerfmt='bo', basefmt='k-')
axes[0].set_title(r'Signal $x[n] = \{1, 2, \mathbf{1}, 1\}$', fontsize=9.5, fontweight='bold', color='darkblue')
axes[0].set_ylabel('Amplitude', fontsize=8.5)
axes[0].grid(True, linestyle='--', alpha=0.6)

axes[1].stem(n_y, y6, linefmt='g-', markerfmt='go', basefmt='k-')
axes[1].set_title(r'Signal $y[n] = \delta[n] - \delta[n-1] + \delta[n-4] + \delta[n-5]$', fontsize=9.5, fontweight='bold', color='darkgreen')
axes[1].set_ylabel('Amplitude', fontsize=8.5)
axes[1].grid(True, linestyle='--', alpha=0.6)

axes[2].stem(n_z6, z6, linefmt='r-', markerfmt='ro', basefmt='k-')
axes[2].set_title(r'Convolution Result $z[n] = x[n] * y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
axes[2].set_xlabel('Index $n$', fontsize=8.5)
axes[2].set_ylabel('Amplitude', fontsize=8.5)
axes[2].grid(True, linestyle='--', alpha=0.6)
plt.show()
```
