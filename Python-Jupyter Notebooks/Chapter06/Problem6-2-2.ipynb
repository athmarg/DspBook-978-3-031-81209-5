```python
import numpy as np
import matplotlib.pyplot as plt

# =========================================================================
# EDUCATIONAL NOTE:
# Since both continuous signals (the original signal with F = 50 Hz and its 
# alias with F_alias = 25 Hz) pass through the exact same sample points when 
# sampled at F_s = 75 Hz, they result in an identical discrete sequence. 
# Therefore, we cannot distinguish between them from the discrete samples alone, 
# which demonstrates the core concept of frequency aliasing for students.
# =========================================================================

# 1. Definition of the continuous signal and parameters
# x(t) = 3 * cos(100 * pi * t) -> F = 50 Hz
A = 3
F = 50.0
omega = 2 * np.pi * F

# Time duration for visualization (e.g., 0.08 seconds to show a few periods)
t_continuous = np.linspace(0, 0.08, 1000)
x_continuous = A * np.cos(omega * t_continuous)

# For part (d): Alias signal having F_alias = 25 Hz -> x_alias(t) = 3 * cos(50 * pi * t)
F_alias = 25.0
omega_alias = 2 * np.pi * F_alias
x_alias_continuous = A * np.cos(omega_alias * t_continuous)

# Sampling frequencies from parts (b) and (c)
Fs_b = 200.0  # Hz
Fs_c = 75.0   # Hz

Ts_b = 1.0 / Fs_b
Ts_c = 1.0 / Fs_c

# Discrete time indices and instants
n_indices_b = np.arange(0, int(0.08 * Fs_b) + 1)
t_samples_b = n_indices_b * Ts_b
x_samples_b = A * np.cos((100 / Fs_b) * np.pi * n_indices_b)

n_indices_c = np.arange(0, int(0.08 * Fs_c) + 1)
t_samples_c = n_indices_c * Ts_c
x_samples_c = A * np.cos((100 / Fs_c) * np.pi * n_indices_c)

# Print mathematical expressions and explanation for verification
print("--- Mathematical Expressions & Aliasing Insight ---")
print(f"Original Signal: x(t) = {A} cos({int(2*F)}pi t) -> F = {F} Hz")
print(f"(b) Sampled at Fs = {Fs_b} Hz: x[n] = 3 cos(n * pi / 2)")
print(f"(c) Sampled at Fs = {Fs_c} Hz: x[n] = 3 cos(4n * pi / 3) = 3 cos(2n * pi / 3)")
print(f"(d) Alias Signal: x_alias(t) = 3 cos(50 * pi * t) -> F_alias = {F_alias} Hz")
print("Note: Because both continuous signals yield the same discrete sequence at Fs = 75 Hz,")
print("they are indistinguishable, clearly illustrating the aliasing phenomenon.")

# 2. Plotting the subplots to visually demonstrate the results
fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# Subplot 1: Part (b) - Fs = 200 Hz (No Aliasing, F < Fs/2)
axes[0].plot(t_continuous * 1000, x_continuous, label=r'$x(t) = 3\cos(100\pi t)$ ($F = 50$ Hz)', color='blue', linewidth=1.5)
axes[0].stem(t_samples_b * 1000, x_samples_b, linefmt='g-', markerfmt='go', basefmt='g--', 
             label=r'Samples at $F_s = 200$ Hz ($x[n] = 3\cos(\frac{\pi}{2}n)$)')
axes[0].set_title(r'Part (b): Sampling at $F_s = 200$ Hz ($F < F_s/2$, No Aliasing)')
axes[0].set_ylabel('Amplitude (V)')
axes[0].axhline(0, color='black', linewidth=0.8, linestyle='--')
axes[0].grid(True, linestyle=':', alpha=0.7)
axes[0].legend(loc='upper center', bbox_to_anchor=(0.5, 1.25), ncol=2)

# Subplot 2: Part (c) & (d) - Fs = 75 Hz (Aliasing, showing both original and alias continuous signals)
axes[1].plot(t_continuous * 1000, x_continuous, label=r'Original $x(t) = 3\cos(100\pi t)$ ($F = 50$ Hz)', color='blue', linewidth=1.5)
axes[1].plot(t_continuous * 1000, x_alias_continuous, label=r'Alias $x_{alias}(t) = 3\cos(50\pi t)$ ($F_{alias} = 25$ Hz)', color='orange', linestyle='--', linewidth=1.5)
axes[1].stem(t_samples_c * 1000, x_samples_c, linefmt='r-', markerfmt='ro', basefmt='r--', 
             label=r'Samples at $F_s = 75$ Hz (Identical for both!)')
axes[1].set_title(r'Parts (c) & (d): Aliasing at $F_s = 75$ Hz ($F > F_s/2$)')
axes[1].set_xlabel('Time (ms)')
axes[1].set_ylabel('Amplitude (V)')
axes[1].axhline(0, color='black', linewidth=0.8, linestyle='--')
axes[1].grid(True, linestyle=':', alpha=0.7)
axes[1].legend(loc='upper center', bbox_to_anchor=(0.5, 1.25), ncol=3)

plt.tight_layout()
plt.show()
```
