```python
# Notebook: Fourier Transform of a Periodic Impulse Train (Dirac Comb)
# Time domain: x(t) = sum_{n} delta(t - nT)
# Frequency domain: X(omega) = (2pi / T) * sum_{k} delta(omega - k * omega_0)

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# ==========================
# Συνάρτηση Οπτικοποίησης
# ==========================
def plot_periodic_impulse_train(T=2.0):
    omega_0 = 2 * np.pi / T  # Θεμελιώδης συχνότητα στη συχνότητα
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 7), sharex=False)
    
    # 1. Time Domain: Ακολουθία κρουστικών (Dirac Comb)
    t_vals = np.arange(-3 * T, 3 * T + 0.001, T)
    peak_time = 1.0
    
    markerline1, stemlines1, baseline1 = ax1.stem(
        t_vals, [peak_time] * len(t_vals), linefmt='b-', markerfmt='bo', basefmt=" "
    )
    plt.setp(stemlines1, linewidth=2)
    
    # Χρήση raw string (r"...") για να μην υπάρχουν warnings escape sequence
    ax1.set_title(r"Periodic Impulse Train ($T$ = %.1f s, $\omega_0$ = %.2f rad/s)" % (T, omega_0), fontsize=12)
    ax1.set_ylabel(r"$x(t) = \sum \delta(t - nT)$", fontsize=11)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.set_xlim(-3 * T - 1, 3 * T + 1)
    ax1.set_ylim(-0.2, 2.0)  # Αυξημένο ύψος άξονα y για να χωράει ο «αέρας» επάνω
    ax1.axhline(0, color='black', linewidth=0.8, linestyle='--')
    
    # Βελάκι και λεζάντα περιόδου T: Τοποθετημένα καθαρά ΠΑΝΩ από τις κορυφές (στο ύψος 1.3 έως 1.7)
    ax1.annotate('', xy=(0, 1.4), xytext=(T, 1.4),
                 arrowprops=dict(arrowstyle='<->', color='blue', lw=1.5))
    ax1.text(T / 2, 1.5, r"$T = %.1f$" % T, ha='center', va='bottom', fontsize=11, color='blue')

    # 2. Frequency Domain: Ακολουθία κρουστικών στη συχνότητα
    omega_vals = np.arange(-4 * omega_0, 4 * omega_0 + 0.001, omega_0)
    amplitude_freq = 2 * np.pi / T
    
    markerline2, stemlines2, baseline2 = ax2.stem(
        omega_vals, [amplitude_freq] * len(omega_vals), linefmt='r-', markerfmt='ro', basefmt=" "
    )
    plt.setp(stemlines2, linewidth=2)
    
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"$\mathcal{X}(j\omega) = \frac{2\pi}{T}\sum \delta(\omega - k\omega_0)$", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.set_xlim(-4 * omega_0 - 1, 4 * omega_0 + 1)
    
    # Δυναμικό όριο y ώστε να υπάρχει χώρος πάνω από την ψηλότερη κορυφή για το βελάκι
    top_ylim = max(6.0, amplitude_freq + 2.0)
    ax2.set_ylim(-0.5, top_ylim)
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    
    # Βελάκι και λεζάντα απόστασης omega_0 στη συχνότητα: Πάνω από τις κορυφές
    arrow_y = amplitude_freq + 0.5
    ax2.annotate('', xy=(0, arrow_y), xytext=(omega_0, arrow_y),
                 arrowprops=dict(arrowstyle='<->', color='red', lw=1.5))
    ax2.text(omega_0 / 2, arrow_y + 0.1, r"$\omega_0 = \frac{2\pi}{T} = %.2f$" % omega_0, ha='center', va='bottom', fontsize=11, color='red')

    plt.tight_layout()
    plt.show()
    
    print(f"Observation (Period T = {T:.1f}):")
    print(f"-> Time Domain: Impulses are spaced by T = {T:.1f} seconds.")
    print(f"-> Frequency Domain: Fourier transform is also an impulse train, but spaced by omega_0 = 2*pi/T = {omega_0:.2f} rad/s")
    print(f"-> The amplitude of each frequency impulse is scaled by 2*pi/T = {amplitude_freq:.2f}.")

# ==========================
# Εφαρμογή Interactive Widget
# ==========================
interact(
    plot_periodic_impulse_train, 
    T=FloatSlider(value=2.0, min=0.8, max=5.0, step=0.5, description="Period T", style={'description_width': 'initial'})
);
```
