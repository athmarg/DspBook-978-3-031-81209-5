```python
# Notebook: Fourier Reconstruction of a Triangular Signal
# Author: Adapted for Springer-style presentation
# Fourier expansion:
# x(t)=E/2 + (4E/pi^2) sum cos((2k+1)t)/(2k+1)^2

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import BoundedIntText, FloatSlider, HBox, VBox, Output
from IPython.display import display


# ==========================
# Parameters
# ==========================

t = np.linspace(-np.pi, np.pi, 2000)


# ==========================
# Triangular signal
# ==========================

def x(t, E=1):
    return np.where(
        t < 0,
        E + E*t/np.pi,
        E - E*t/np.pi
    )


# ==========================
# Fourier expansion
# ==========================

def x_fourier(t, N, E=1):

    xf = E/2

    for k in range(N):
        n = 2*k + 1
        xf += (4*E/np.pi**2)*np.cos(n*t)/n**2

    return xf


# ==========================
# Controls
# ==========================

N = BoundedIntText(
    value=1,
    min=1,
    max=20,
    description="N"
)

E = FloatSlider(
    value=1,
    min=0.1,
    max=5,
    step=0.1,
    description="E"
)

out = Output()


# ==========================
# Plot
# ==========================

def plot(*args):

    with out:

        out.clear_output(wait=True)

        xx = x(t, E.value)
        xf = x_fourier(t, N.value, E.value)

        fig, ax = plt.subplots(figsize=(9, 4))

        ax.plot(
            t, xx,
            'r',
            linewidth=3,
            label="Original signal"
        )

        ax.plot(
            t, xf,
            'b',
            linewidth=2,
            label=f"Fourier expansion (N={N.value})"
        )

        ax.set_xlabel(r"$\omega t$")
        ax.set_ylabel("Amplitude")
        ax.grid(True)
        ax.legend()

        ax.set_title(
            "Triangular Signal Fourier Reconstruction"
        )

        display(fig)
        plt.close(fig)


N.observe(plot, names="value")
E.observe(plot, names="value")


display(
    VBox(
        [
            HBox([N, E]),
            out
        ]
    )
)

# Καλώ την plot μία φορά αμέσως για να αρχικοποιηθεί το Output widget
plot()
```
