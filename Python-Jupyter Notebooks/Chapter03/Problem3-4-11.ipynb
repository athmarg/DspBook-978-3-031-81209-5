```python
# Notebook: Inverse Fourier Transform via Partial Fraction Expansion
# Function to invert: X(j\omega) = \sin(2\omega) / (\omega^2 + 8\omega + 20)

import sympy as sp
import numpy as np

# Define real symbols for symbolic math
t_sym, omega_sym = sp.symbols('t omega', real=True)

print("--- 1. SYMBOLIC SETUP & PARTIAL FRACTION EXPANSION ---")
# Let's define the core rational fraction Y(j\omega) = 1 / ((\omega - 10)(\omega - 2)) as shown in the slide
Y_omega = 1 / ((omega_sym - 10) * (omega_sym - 2))
partial_fractions = sp.apart(Y_omega, omega_sym)

print("Partial fraction expansion of 1/((omega - 10)*(omega - 2)):")
display(partial_fractions)


# ==========================================================
# 2. NUMERICAL VERIFICATION & TIME-DOMAIN PLOTTING
# ==========================================================
import matplotlib.pyplot as plt

def plot_time_response():
    # Define time vector using NumPy arrays
    t_vals = np.linspace(-6, 6, 600)
    
    # Sign function definition using NumPy
    sgn = lambda val: np.where(val > 0, 1.0, np.where(val < 0, -1.0, 0.0))
    
    # Component 1 based on time shift (t + 2) -> using t_vals array
    arg1_plus = t_vals + 2
    x1_val = (1/32) * (np.exp(10j * arg1_plus) - np.exp(2j * arg1_plus)) * sgn(arg1_plus)
    
    # Component 2 based on time shift (t - 2) -> using t_vals array
    arg1_minus = t_vals - 2
    x2_val = (1/32) * (np.exp(10j * arg1_minus) - np.exp(2j * arg1_minus)) * sgn(arg1_minus)
    
    # Total response (taking real part of the inverse transform components)
    x_total = np.real(x1_val - x2_val)
    
    fig, ax = plt.subplots(figsize=(10, 4.5))
    ax.plot(t_vals, x_total, 'b-', linewidth=2, label=r"$x(t) = x_1(t) - x_2(t)$")
    
    ax.set_title(r"Time-Domain Response $x(t)$ via Partial Fractions & Shift Properties", fontsize=12)
    ax.set_xlabel(r"Time $t$ (s)", fontsize=11)
    ax.set_ylabel(r"$x(t)$", fontsize=11)
    ax.grid(True, linestyle=":", alpha=0.7)
    ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax.axvline(0, color='black', linewidth=0.8, linestyle='--')
    
    # Move legend outside to the top right to avoid covering the curves
    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1), fontsize=10)
    
    plt.tight_layout()
    plt.show()

# Execute verification plot
plot_time_response()
```
