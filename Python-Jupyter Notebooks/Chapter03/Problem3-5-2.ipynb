```python
# Notebook: System Frequency Response from Input/Output and Impulse Response
# Input: x(t) = (e^{-t} + e^{-3t})u(t)
# Output: y(t) = (2e^{-t} - 2e^{-4t})u(t)
# Goal: Compute H(j\omega) = Y(j\omega)/X(j\omega), h(t), and plot amplitude/phase responses.

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Define real symbols
t, omega = sp.symbols('t omega', real=True)

print("--- 1. SYMBOLIC TRANSFORMS X(j\\omega), Y(j\\omega) AND H(j\\omega) ---")
# Define x(t) and y(t) using Heaviside step function
x_t = (sp.exp(-t) + sp.exp(-3*t)) * sp.Heaviside(t)
y_t = (2*sp.exp(-t) - 2*sp.exp(-4*t)) * sp.Heaviside(t)

# Compute Fourier transforms
X_omega = sp.fourier_transform(x_t, t, omega)
Y_omega = sp.fourier_transform(y_t, t, omega)

# Frequency response H(j\omega) = Y(j\omega) / X(j\omega)
H_omega = sp.simplify(Y_omega / X_omega)

print("\nDerived Frequency Response H(j\\omega):")
display(H_omega)


# ==========================================================
# 2. NUMERICAL EVALUATION & PLOTTING (Amplitude & Phase Responses)
# ==========================================================
def plot_system_responses():
    # Define frequency vector omega from -15 to 15
    omega_vals = np.linspace(-15, 15, 1000)
    
    with np.errstate(divide='ignore', invalid='ignore'):
        # Evaluate H(j\omega) numerically using the analytical expression from the slide:
        # H(j\omega) = (9 + 3*j*omega) / (8 + 6*j*omega - omega**2)
        # Note: In Python/NumPy, complex numbers use 'j' or '1j'
        H_vals = (9 + 3 * 1j * omega_vals) / (8 + 6 * 1j * omega_vals - omega_vals**2)
        
        amplitude_response = np.abs(H_vals)
        phase_response = np.angle(H_vals)

    # Create figure with two subplots: Amplitude Response and Phase Response
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 7.5), sharex=True)
    
    # Plot Amplitude Response
    ax1.plot(omega_vals, amplitude_response, 'b-', linewidth=2, label=r"$|\mathcal{H}(j\omega)| = \left|\frac{9+3j\omega}{8+6j\omega-\omega^2}\right|$")
    ax1.set_title(r"System Amplitude and Phase Responses", fontsize=13)
    ax1.set_ylabel(r"Amplitude $|\mathcal{H}(j\omega)|$", fontsize=11)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax1.legend(loc='lower right', fontsize=10)
    
    # Plot Phase Response (Legend placed at the lower right)
    ax2.plot(omega_vals, phase_response, 'r-', linewidth=2, label=r"$\angle \mathcal{H}(j\omega) = \text{atan2}(\Im\{\mathcal{H}\}, \Re\{\mathcal{H}\})$")
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"Phase $\angle \mathcal{H}(j\omega)$ (rad)", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax2.axvline(0, color='black', linewidth=0.8, linestyle='--')
    
    # Positioned legend at the lower right of the bottom subplot
    ax2.legend(loc='lower left', fontsize=10)
    
    plt.tight_layout()
    plt.show()

# Execute plotting function
plot_system_responses()
```
