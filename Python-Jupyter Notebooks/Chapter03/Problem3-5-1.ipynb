```python
# Notebook: Frequency Response, Amplitude, and Phase Analysis
# Impulse response: h(t) = u(t - 2) - u(t + 1)
# Goal: Compute H(j\omega), plot amplitude response |H(j\omega)| and phase response \angle H(j\omega)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Define real symbols
t, omega = sp.symbols('t omega', real=True)

print("--- 1. SYMBOLIC FOURIER TRANSFORM OF h(t) ---")
# Define h(t) using Heaviside step function: Heaviside(t)
# h(t) = u(t - 2) - u(t + 1)
h_t = sp.Heaviside(t - 2) - sp.Heaviside(t + 1)
display("Impulse response h(t):", h_t)

# Compute the frequency response H(j\omega)
H_omega_direct = sp.integrate(h_t * sp.exp(-sp.I * omega * t), (t, -sp.oo, sp.oo))
H_omega_simplified = sp.simplify(H_omega_direct)

print("\nDerived frequency response H(j\\omega):")
display(H_omega_simplified)


# ==========================================================
# 2. NUMERICAL EVALUATION & PLOTTING (Amplitude & Phase Responses)
# ==========================================================
def plot_amplitude_and_phase_responses():
    # Define frequency vector omega from -15 to 15
    omega_vals = np.linspace(-15, 15, 1000)
    
    with np.errstate(divide='ignore', invalid='ignore'):
        # Analytical H(j\omega) evaluation based on the system's frequency response
        H_vals = -3 * np.exp(-0.5j * omega_vals) * np.sin(1.5 * omega_vals) / omega_vals
        H_vals[omega_vals == 0] = 3.0
        
        amplitude_response = np.abs(H_vals)
        phase_response = np.angle(H_vals)

    # Create figure with two subplots: Amplitude Response and Phase Response
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 7.5), sharex=True)
    
    # Plot Amplitude Response
    ax1.plot(omega_vals, amplitude_response, 'b-', linewidth=2, label=r"$|\mathcal{H}(j\omega)| = 3 \left|\text{sinc}\left(\frac{3\omega}{2\pi}\right)\right|$")
    ax1.set_title(r"System Frequency Response: Amplitude and Phase", fontsize=13)
    ax1.set_ylabel(r"Amplitude $|\mathcal{H}(j\omega)|$", fontsize=11)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax1.legend(loc='upper right', fontsize=10)
    
    # Plot Phase Response (Legend placed at the bottom/lower-left to avoid overlapping the curves)
    ax2.plot(omega_vals, phase_response, 'r-', linewidth=2, label=r"$\angle \mathcal{H}(j\omega) = -\frac{\omega}{2}$")
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"Phase $\angle \mathcal{H}(j\omega)$ (rad)", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax2.axvline(0, color='black', linewidth=0.8, linestyle='--')
    
    # Positioned legend at the lower center/left of the bottom subplot
    ax2.legend(loc='lower right', fontsize=10)
    
    plt.tight_layout()
    plt.show()

# Execute plotting function
plot_amplitude_and_phase_responses()
```
