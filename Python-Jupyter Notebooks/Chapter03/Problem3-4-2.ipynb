```python
# Notebook: Inverse Fourier Transform using SymPy (Symbolic Calculation)
# Function to invert: X(j\omega) = cos(4\omega + \pi/3)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Define mathematical symbols
t, omega = sp.symbols('t omega', real=True)

# Define the frequency-domain function
# X(j\omega) = cos(4*omega + pi/3)
X_omega = sp.cos(4 * omega + sp.pi / 3)

print("--- 1. SYMBOLIC INVERSE FOURIER TRANSFORM (SymPy) ---")
print("Given frequency-domain function X(j\\omega):")
display(X_omega)

# The inverse Fourier transform formula:
# x(t) = (1 / 2\pi) * \int_{-\infty}^{+\infty} X(j\omega) e^{j\omega t} d\omega
# SymPy evaluates this integral analytically using Dirac delta distributions.

# Calculate the integral from -oo to +oo
integrand = X_omega * sp.exp(sp.I * omega * t)
integral_result = sp.integrate(integrand, (omega, -sp.oo, sp.oo))
x_t_symbolic = (1 / (2 * sp.pi)) * integral_result

print("\nDerived inverse Fourier transform x(t):")
display(x_t_symbolic)


# ==========================================================
# 2. VISUALIZATION OF THE TIME-DOMAIN IMPULSES
# ==========================================================
# Since Dirac impulses cannot be plotted directly with infinite height,
# we represent them using stem plots at t = -4 and t = 4.

def plot_inverse_transform():
    # Impulse positions
    t_positions = [-4.0, 4.0]
    
    # Compute complex amplitudes: e^(j*pi/3)/(4*pi) and e^(-j*pi/3)/(4*pi)
    amp_neg4 = sp.exp(sp.I * sp.pi / 3) / (4 * sp.pi)
    amp_pos4 = sp.exp(-sp.I * sp.pi / 3) / (4 * sp.pi)
    
    # Use the magnitude of the complex amplitudes for the stem heights
    magnitudes = [float(sp.Abs(amp_neg4)), float(sp.Abs(amp_pos4))]
    
    fig, ax = plt.subplots(figsize=(10, 4))
    
    markerline, stemlines, baseline = ax.stem(
        t_positions, magnitudes, linefmt='b-', markerfmt='bo', basefmt=" "
    )
    plt.setp(stemlines, linewidth=2.5)
    
    ax.set_title(r"Inverse Fourier Transform Result: $x(t)$ as Impulses at $t = \pm 4$", fontsize=12)
    ax.set_xlabel(r"Time $t$ (s)", fontsize=11)
    ax.set_ylabel(r"Amplitude Magnitude of $\delta(t \pm 4)$", fontsize=11)
    ax.grid(True, linestyle=":", alpha=0.7)
    ax.set_xlim(-7, 7)
    ax.set_ylim(0, 0.09)
    ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
    
    # Position the text labels clearly to the left and right of the respective lines
    # Left impulse at t = -4: place text to its left
    ax.text(-4.8, 0.025, r"$\frac{e^{j\pi/3}}{4\pi} \delta(t+4)$", ha='right', va='center', fontsize=11, color='blue')
    # Right impulse at t = 4: place text to its right
    ax.text(4.8, 0.025, r"$\frac{e^{-j\pi/3}}{4\pi} \delta(t-4)$", ha='left', va='center', fontsize=11, color='blue')
    
    plt.tight_layout()
    plt.show()

# Execute the visualization function
plot_inverse_transform()
```
