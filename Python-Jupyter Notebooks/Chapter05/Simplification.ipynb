```python
# Notebook: Clean Symbolic Block Diagram Reduction with True Fractions
import sympy as sp

# Ενεργοποίηση ποιοτικής εκτύπωσης unicode/math για καθαρά κλάσματα
sp.init_printing(use_unicode=True)

# 1. Ορισμός συμβόλων
s = sp.symbols('s')
G1 = sp.Symbol('G_1', commutative=False)
G2 = sp.Symbol('G_2', commutative=False)
G3 = sp.Symbol('G_3', commutative=False)
G4 = sp.Symbol('G_4', commutative=False)
H1 = sp.Symbol('H_1', commutative=False)
H2 = sp.Symbol('H_2', commutative=False)

# 2. Συμβολική μείωση βήμα-βήμα
G23 = G2 + G3
G234 = G23 * G4
inner_loop = G234 / (1 + G234 * H1)
forward_path = G1 * inner_loop
H_total = forward_path / (1 + forward_path * H2)

# Χρήση cancel() και together() για καθαρή μορφοποίηση κλασμάτων
print("--- Step 1: Parallel Connection (G2 + G3) ---")
display(G23)

print("\n--- Step 2: Series Connection with G4 ---")
display(G234)

print("\n--- Step 3 & 4: Inner Loop Feedback with H1 ---")
display(sp.cancel(inner_loop))

print("\n--- Step 5: Series Connection with G1 ---")
display(sp.cancel(forward_path))

print("\n--- Final Equivalent Transfer Function H_total(s) as a Fraction ---")
display(sp.cancel(H_total))
```
