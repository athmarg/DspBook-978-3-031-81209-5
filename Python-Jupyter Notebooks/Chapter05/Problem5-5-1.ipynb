```python
# Notebook: Step Response Analysis & Partial Fraction Expansion (Example)
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# 1. Define symbolic variable s and time t
s = sp.Symbol('s')
t_sym = sp.Symbol('t', positive=True)

# 2. Define Transfer Function H(s) = (s + 5) / (s^2 + 3s + 2)
num_expr = s + 5
den_expr = s**2 + 3*s + 2  # factors as (s+1)(s+2)
H_sym = num_expr / den_expr

# 3. Step Response in Laplace domain: Y(s) = H(s) / s
Y_s = H_sym / (1 / s) # Wait, Y(s) = H(s) * (1/s)
Y_s = H_sym / s

# Partial fraction expansion to find analytical coefficients (5/2, -4, 3/2)
pfe = sp.apart(Y_s, s)

# Inverse Laplace transform to get y(t) analytically
y_t = sp.inverse_laplace_transform(Y_s, s, t_sym)

print("=== Symbolic Step Response Analysis ===")
print(f"• Transfer Function H(s): {H_sym}")
print(f"• Y(s) = H(s)/s expanded: {pfe}")
print(f"• Analytical Step Response y(t): {y_t}")
print("-" * 50)

# 4. Control Library Verification & Simulation
num_coeffs = [1, 5]
den_coeffs = [1, 3, 2]
sys = ct.tf(num_coeffs, den_coeffs)

# Compute step response using control library
t = np.linspace(0, 6, 500)
t_out, y_response = ct.step_response(sys, T=t)

# Final Value Theorem verification: H(0) = 5/2 = 2.5
h_zero = ct.evalfr(sys, 0)
print(f"• Final Value Theorem Check H(0) = {h_zero:.2f}")

# 5. Plotting the Step Response, Transient, and Steady-State
plt.figure(figsize=(8, 4.5))
plt.plot(t_out, y_response, color='darkblue', lw=2.5, label='Step Response $y(t)$')
plt.axhline(2.5, color='red', linestyle='--', label='Steady-State $y_{ss}(t) = 2.5$')
plt.title(r'Step Response of $\mathcal{H}(s) = \dfrac{s+5}{s^2+3s+2}$', fontsize=11)
plt.xlabel('Time $t$ (s)', fontsize=11)
plt.ylabel('Amplitude $y(t)$', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='best', frameon=True, fontsize=9)
plt.show()
```
