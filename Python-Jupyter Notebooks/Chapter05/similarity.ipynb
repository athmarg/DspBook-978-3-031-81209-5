```python
# Notebook: Verification of Transfer Function Invariance under Similarity Transformations
import sympy as sp

# 1. Definition of Original System Matrices (A, B, C, D)
s = sp.symbols('s')
A = sp.Matrix([[0, 2], [-1, -3]])
B = sp.Matrix([0, 1])
C = sp.Matrix([[2, 3]])
D = sp.Matrix([[-1]])
I = sp.eye(2)

print("--- Original System Matrices ---")
print("Matrix A:")
display(A)
print("Matrix B:")
display(B)
print("Matrix C:")
display(C)
print("Matrix D:")
display(D)

# Original Transfer Function H(s) = C * (sI - A)^-1 * B + D
H_original = C * (s * I - A).inv() * B + D
H_original_simplified = sp.simplify(H_original[0, 0])
print("\n--- Original Transfer Function H(s) ---")
display(H_original_simplified)

# 2. Definition of Transformation Matrices W1, W2, W3
W1 = sp.Matrix([[5, 1], [0, 6]])
W2 = sp.Matrix([[-4, 0], [8, 6]])
W3 = sp.Matrix([[5, 0], [15, 3]])

transformations = [("W1", W1), ("W2", W2), ("W3", W3)]

# 3. Verification of Invariance for each Transformation W_i
for name, W in transformations:
    print(f"\n--- Testing Transformation Matrix {name} ---")
    print(f"Matrix {name}:")
    display(W)
    
    # Compute inverse of transformation matrix
    W_inv = W.inv()
    
    # Compute transformed state-space matrices
    A_tilde = W * A * W_inv
    B_tilde = W * B
    C_tilde = C * W_inv
    
    print(f"Transformed Matrix A_tilde ({name}):")
    display(A_tilde)
    print(f"Transformed Matrix B_tilde ({name}):")
    display(B_tilde)
    print(f"Transformed Matrix C_tilde ({name}):")
    display(C_tilde)
    
    # Compute transformed transfer function H_tilde(s)
    H_tilde = C_tilde * (s * I - A_tilde).inv() * B_tilde + D
    H_tilde_simplified = sp.simplify(H_tilde[0, 0])
    
    print(f"Transformed Transfer Function for {name}:")
    display(H_tilde_simplified)
    
    # Check if it matches the original transfer function
    diff = sp.simplify(H_tilde_simplified - H_original_simplified)
    print(f"Difference (H_tilde - H_original) for {name}:", diff)
```
