```python
# Notebook: Solution of Exercise using Python Control Library
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

import control as ct

# 1. Define the individual transfer functions
# H1(s) = 2 / (s + 3)
H1 = ct.tf([2], [1, 3])

# H2(s) = (2s + 5) / (s^2 - 5s + 7)
H2 = ct.tf([2, 5], [1, -5, 7])

# H3(s) = (s - 2) / (s^3 - 2s^2 + 25s - 50)
H3 = ct.tf([1, -2], [1, -2, 25, -50])

# --- (a) Series and Parallel Connections ---
# Series connection of H1, H2, and H3 (G_series = H1 * H2 * H3)
G_series = ct.series(H1, H2, H3)

# Parallel connection of H1, H2, and H3 (G_parallel = H1 + H2 + H3)
G_parallel = ct.parallel(H1, H2, H3)

print("--- (a) Series Connection ---")
print(G_series)

print("\n--- (a) Parallel Connection ---")
print(G_parallel)

# --- (b) Negative Feedback Connection of H1 and H2 ---
# Assuming H1 is in the forward path and H2 is in the feedback path: ct.feedback(H1, H2)
G_feedback = ct.feedback(H1, H2)

print("\n--- (b) Negative Feedback Connection (H1 and H2) ---")
print(G_feedback)
```
