```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from ipywidgets import Dropdown, FloatSlider, interactive
from IPython.display import display, Markdown

# ------------------------------------------------------------
# Function of a Random Variable
# Y = g(X)
# ------------------------------------------------------------

def random_variable_transformation(transformation, a, b):

    rng = np.random.default_rng(12345)

    N = 200000

    # Input random variable
    X = rng.normal(0, 1, N)

    if transformation == 'Y = aX + b':
        Y = a*X + b
        transformation_text = f'Y = {a:.2f} X + {b:.2f}'

    elif transformation == 'Y = X²':
        Y = X**2
        transformation_text = 'Y = X²'

    elif transformation == 'Y = |X|':
        Y = np.abs(X)
        transformation_text = 'Y = |X|'

    elif transformation == 'Y = exp(aX)':
        Y = np.exp(a*X)
        transformation_text = f'Y = exp({a:.2f} X)'

    # --------------------------------------------------------
    # Input PDF
    # --------------------------------------------------------

    x = np.linspace(-4, 4, 1000)
    pdf_x = norm.pdf(x)

    # --------------------------------------------------------
    # Output PDF estimated from samples
    # --------------------------------------------------------

    low = np.percentile(Y, 0.2)
    high = np.percentile(Y, 99.8)

    bins = np.linspace(low, high, 120)

    fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

    # Input PDF
    axes[0].plot(x, pdf_x, linewidth=2)

    axes[0].set_title('Input Random Variable X')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('$f_X(x)$')
    axes[0].grid(True, alpha=0.3)

    # Transformation
    x_transform = np.linspace(-3, 3, 1000)

    if transformation == 'Y = aX + b':
        y_transform = a*x_transform + b

    elif transformation == 'Y = X²':
        y_transform = x_transform**2

    elif transformation == 'Y = |X|':
        y_transform = np.abs(x_transform)

    elif transformation == 'Y = exp(aX)':
        y_transform = np.exp(a*x_transform)

    axes[1].plot(
        x_transform,
        y_transform,
        linewidth=2
    )

    axes[1].set_title(f'Transformation\n{transformation_text}')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('y = g(x)')
    axes[1].grid(True, alpha=0.3)

    # Output PDF
    axes[2].hist(
        Y,
        bins=bins,
        density=True,
        alpha=0.6
    )

    axes[2].set_title('Output Random Variable Y')
    axes[2].set_xlabel('y')
    axes[2].set_ylabel('$f_Y(y)$')
    axes[2].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
    plt.close(fig)

    print(f'Transformation: {transformation_text}')
    print()
    print(f'E[X]       ≈ {np.mean(X):.6f}')
    print(f'Var(X)     ≈ {np.var(X):.6f}')
    print()
    print(f'E[Y]       ≈ {np.mean(Y):.6f}')
    print(f'Var(Y)     ≈ {np.var(Y):.6f}')


display(Markdown("## Function of a Random Variable — Y = g(X)"))

display(
    interactive(
        random_variable_transformation,

        transformation=Dropdown(
            options=[
                'Y = aX + b',
                'Y = X²',
                'Y = |X|',
                'Y = exp(aX)'
            ],
            value='Y = aX + b',
            description='Function:'
        ),

        a=FloatSlider(
            value=1,
            min=-2,
            max=2,
            step=0.1,
            description='a:',
            continuous_update=True
        ),

        b=FloatSlider(
            value=0,
            min=-3,
            max=3,
            step=0.1,
            description='b:',
            continuous_update=True
        )
    )
)
```
