```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    Checkbox,
    HTML,
    VBox,
    HBox,
    Layout
)
from IPython.display import display

# ============================================================
# TRIGONOMETRIC FUNCTIONS:
# SINE, COSINE AND THE UNIT CIRCLE
# ============================================================

plt.ioff()

# ============================================================
# CSS
#
# IMPORTANT:
# The rules below suppress horizontal scrolling in the
# Jupyter/Binder output area and force widget containers
# to respect the available width.
# ============================================================

display(HTML("""
<style>

/* ------------------------------------------------------------
   Notebook-specific text styles
   ------------------------------------------------------------ */

.trig-title {
    font-family: Arial, sans-serif;
    font-size: 19px;
    font-weight: bold;
    color: #0b3d91;
    margin-bottom: 8px;
}

.trig-small-title {
    font-family: Arial, sans-serif;
    font-size: 16px;
    font-weight: bold;
    margin-bottom: 6px;
}

.trig-label {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
}

.trig-value {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
    color: #0b3d91;
}

/* ------------------------------------------------------------
   REMOVE HORIZONTAL OUTPUT SCROLLBARS
   Classic Notebook + JupyterLab + Binder
   ------------------------------------------------------------ */

.output_scroll,
.output_area,
.output_subarea,
.widget-subarea,
.jupyter-widgets-output-area,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    overflow-x: hidden !important;
    max-width: 100% !important;
}

/* Prevent the old classic-notebook output_scroll box */
.output_scroll {
    box-shadow: none !important;
    height: auto !important;
    overflow-y: visible !important;
}

/* Widget boxes must include padding/border in their width */
.widget-box,
.jupyter-widgets {
    box-sizing: border-box !important;
}

/* Prevent canvas elements from forcing output width */
.jupyter-widgets canvas {
    max-width: 100% !important;
}

</style>
"""))

# ============================================================
# FIXED PARAMETERS
# ============================================================

T_MAX = 2.0
N_POINTS = 1500

t_axis = np.linspace(
    0.0,
    T_MAX,
    N_POINTS
)

# ============================================================
# SLIDER STYLE
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='175px'
)

value_layout = Layout(
    width='48px',
    min_width='48px',
    margin='0px 0px 0px 5px'
)

label_layout = Layout(
    width='110px',
    min_width='110px'
)

row_layout = Layout(
    width='330px',
    height='36px',
    min_height='36px',
    align_items='center'
)

# ============================================================
# SLIDERS
# ============================================================

amplitude_slider = FloatSlider(
    min=0.10,
    max=3.00,
    step=0.05,
    value=1.00,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

frequency_slider = FloatSlider(
    min=0.10,
    max=3.00,
    step=0.05,
    value=1.00,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

phase_slider = FloatSlider(
    min=-np.pi,
    max=np.pi,
    step=np.pi / 36.0,
    value=0.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

time_slider = FloatSlider(
    min=0.0,
    max=T_MAX,
    step=0.01,
    value=0.75,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

show_values_checkbox = Checkbox(
    value=True,
    description='Show values at current time',
    indent=False,
    layout=Layout(
        width='250px',
        margin='4px 0px 0px 110px'
    )
)

# ============================================================
# CURRENT VALUE LABELS
# ============================================================

amplitude_value = HTML(
    '<div class="trig-value">1.00</div>',
    layout=value_layout
)

frequency_value = HTML(
    '<div class="trig-value">1.00</div>',
    layout=value_layout
)

phase_value = HTML(
    '<div class="trig-value">0.00</div>',
    layout=value_layout
)

time_value = HTML(
    '<div class="trig-value">0.75</div>',
    layout=value_layout
)

# ============================================================
# PARAMETER LABELS
# ============================================================

amplitude_label = HTML(
    '<div class="trig-label">Amplitude A:</div>',
    layout=label_layout
)

frequency_label = HTML(
    '<div class="trig-label">Frequency f:</div>',
    layout=label_layout
)

phase_label = HTML(
    '<div class="trig-label">Phase φ:</div>',
    layout=label_layout
)

time_label = HTML(
    '<div class="trig-label" style="color:#16802b;">Time t:</div>',
    layout=label_layout
)

# ============================================================
# PARAMETER ROWS
# ============================================================

amplitude_row = HBox(
    [
        amplitude_label,
        amplitude_slider,
        amplitude_value
    ],
    layout=row_layout
)

frequency_row = HBox(
    [
        frequency_label,
        frequency_slider,
        frequency_value
    ],
    layout=row_layout
)

phase_row = HBox(
    [
        phase_label,
        phase_slider,
        phase_value
    ],
    layout=row_layout
)

time_row = HBox(
    [
        time_label,
        time_slider,
        time_value
    ],
    layout=row_layout
)

# ============================================================
# PARAMETER PANEL
# ============================================================

parameter_panel = VBox(
    [
        HTML(
            '<div class="trig-title">Signal Parameters</div>'
        ),
        amplitude_row,
        frequency_row,
        phase_row,
        time_row,
        show_values_checkbox
    ],
    layout=Layout(
        width='350px',
        padding='9px 10px',
        border='1px solid #c7d3df',
        overflow='hidden'
    )
)

# ============================================================
# VALUES AT CURRENT TIME
# ============================================================

current_values = HTML()

current_values_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:18px;
            font-weight:bold;
            color:#19752a;
            margin-bottom:6px;
        ">
        Values at Current Time
        </div>
        """),
        current_values
    ],
    layout=Layout(
        width='350px',
        padding='9px 10px',
        border='1px solid #b8dabd',
        margin='7px 0px 0px 0px',
        overflow='hidden'
    )
)

# ============================================================
# RANGE PANEL
# ============================================================

range_panel = HTML("""
<div style="
    width:330px;
    padding:9px 10px;
    border:1px solid #b9cee9;
    font-family:Arial;
    font-size:14px;
    line-height:1.55;
    margin-top:7px;
    box-sizing:border-box;
">

<div style="
    font-size:17px;
    font-weight:bold;
    color:#1249a0;
    margin-bottom:5px;
">
Ranges for any θ
</div>

<div>
• &nbsp; −1 ≤
<span style="color:#d7191c;"><b>sin(θ)</b></span>
≤ 1
</div>

<div>
• &nbsp; −1 ≤
<span style="color:#1555cc;"><b>cos(θ)</b></span>
≤ 1
</div>

</div>
""")

# ============================================================
# LEFT COLUMN
# ============================================================

left_column = VBox(
    [
        parameter_panel,
        current_values_panel,
        range_panel
    ],
    layout=Layout(
        width='360px',
        min_width='360px',
        overflow='hidden'
    )
)

# ============================================================
# CENTRAL FIGURE:
# SINE AND COSINE WAVES
#
# Figure size and canvas size intentionally correspond.
# This avoids clipping by ipympl.
# ============================================================

fig_wave, ax_wave = plt.subplots(
    figsize=(5.4, 4.55)
)

fig_wave.canvas.header_visible = False

fig_wave.canvas.layout = Layout(
    width='540px',
    height='455px',
    overflow='hidden'
)

ax_wave.set_title(
    'Sine and Cosine Waves',
    fontsize=14,
    fontweight='bold',
    color='#112998'
)

ax_wave.set_xlabel(
    'Time t (s)',
    fontsize=10
)

ax_wave.set_ylabel(
    'Amplitude',
    fontsize=10
)

ax_wave.set_xlim(
    0.0,
    T_MAX
)

ax_wave.set_ylim(
    -3.3,
    3.3
)

ax_wave.grid(
    True,
    linestyle=':',
    alpha=0.45
)

# ============================================================
# INITIAL CURVES
# ============================================================

A0 = amplitude_slider.value
f0 = frequency_slider.value
phi0 = phase_slider.value

omega0 = 2.0 * np.pi * f0

sine0 = A0 * np.sin(
    omega0 * t_axis + phi0
)

cosine0 = A0 * np.cos(
    omega0 * t_axis + phi0
)

line_sine, = ax_wave.plot(
    t_axis,
    sine0,
    color='red',
    linewidth=2.0,
    label=r'$A\sin(\omega t+\phi)$'
)

line_cosine, = ax_wave.plot(
    t_axis,
    cosine0,
    color='blue',
    linewidth=2.0,
    label=r'$A\cos(\omega t+\phi)$'
)

# ============================================================
# CURRENT-TIME MARKER
# ============================================================

time_marker = ax_wave.axvline(
    time_slider.value,
    color='black',
    linestyle='--',
    linewidth=1.2,
    alpha=0.8
)

theta0 = (
    omega0 * time_slider.value
    + phi0
)

point_sine, = ax_wave.plot(
    [time_slider.value],
    [A0 * np.sin(theta0)],
    marker='o',
    markersize=6,
    color='red'
)

point_cosine, = ax_wave.plot(
    [time_slider.value],
    [A0 * np.cos(theta0)],
    marker='o',
    markersize=6,
    color='blue'
)

ax_wave.legend(
    loc='upper right',
    fontsize=8
)

fig_wave.subplots_adjust(
    left=0.12,
    right=0.97,
    top=0.91,
    bottom=0.14
)

# ============================================================
# EQUATIONS PANEL
# ============================================================

equations_html = HTML()

equation_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:16px;
            font-weight:bold;
            color:#1249a0;
            margin-bottom:4px;
        ">
        Equations
        </div>
        """),
        equations_html
    ],
    layout=Layout(
        width='540px',
        padding='7px 10px',
        border='1px solid #b9cee9',
        margin='4px 0px 0px 0px',
        overflow='hidden'
    )
)

# ============================================================
# CENTRAL COLUMN
# ============================================================

wave_box = VBox(
    [
        fig_wave.canvas,
        equation_panel
    ],
    layout=Layout(
        width='555px',
        min_width='555px',
        align_items='center',
        overflow='hidden'
    )
)

# ============================================================
# UNIT CIRCLE FIGURE
#
# Again the physical figure and CSS canvas dimensions agree.
# ============================================================

fig_circle, ax_circle = plt.subplots(
    figsize=(3.35, 3.35)
)

fig_circle.canvas.header_visible = False

fig_circle.canvas.layout = Layout(
    width='335px',
    height='335px',
    overflow='hidden'
)

ax_circle.set_title(
    'Trigonometric Circle (Unit Circle)',
    fontsize=12,
    fontweight='bold',
    color='#6a1b9a'
)

ax_circle.set_aspect(
    'equal',
    adjustable='box'
)

ax_circle.set_xlim(
    -1.35,
    1.35
)

ax_circle.set_ylim(
    -1.35,
    1.35
)

ax_circle.set_xticks(
    [-1, 0, 1]
)

ax_circle.set_yticks(
    [-1, 0, 1]
)

ax_circle.axhline(
    0,
    color='black',
    linewidth=1.0
)

ax_circle.axvline(
    0,
    color='black',
    linewidth=1.0
)

# ============================================================
# UNIT CIRCLE
# ============================================================

circle_angle = np.linspace(
    0,
    2.0 * np.pi,
    500
)

ax_circle.plot(
    np.cos(circle_angle),
    np.sin(circle_angle),
    color='black',
    linewidth=1.5
)

# ============================================================
# RADIUS OP
# ============================================================

radius_line, = ax_circle.plot(
    [0, np.cos(theta0)],
    [0, np.sin(theta0)],
    color='purple',
    linewidth=2.0
)

# ============================================================
# POINT P
# ============================================================

circle_point, = ax_circle.plot(
    [np.cos(theta0)],
    [np.sin(theta0)],
    marker='o',
    markersize=7,
    color='red'
)

# ============================================================
# PROJECTIONS
# ============================================================

cos_projection, = ax_circle.plot(
    [0, np.cos(theta0)],
    [0, 0],
    color='blue',
    linestyle='--',
    linewidth=1.6
)

sin_projection, = ax_circle.plot(
    [np.cos(theta0), np.cos(theta0)],
    [0, np.sin(theta0)],
    color='red',
    linestyle='--',
    linewidth=1.6
)

# ============================================================
# ANGLE ARC
# ============================================================

angle_arc, = ax_circle.plot(
    [],
    [],
    color='purple',
    linewidth=1.7
)

# ============================================================
# TEXT LABELS
# ============================================================

theta_text = ax_circle.text(
    0.20,
    0.15,
    r'$\theta$',
    fontsize=12,
    color='purple'
)

point_text = ax_circle.text(
    np.cos(theta0) + 0.06,
    np.sin(theta0) + 0.06,
    'P',
    fontsize=11,
    color='red'
)

ax_circle.text(
    1.16,
    -0.10,
    'x',
    fontsize=10
)

ax_circle.text(
    -0.10,
    1.16,
    'y',
    fontsize=10
)

fig_circle.subplots_adjust(
    left=0.14,
    right=0.95,
    top=0.87,
    bottom=0.10
)

# ============================================================
# UNIT-CIRCLE VALUES PANEL
# ============================================================

circle_values_html = HTML()

circle_values_panel = VBox(
    [
        circle_values_html
    ],
    layout=Layout(
        width='330px',
        padding='4px 6px',
        overflow='hidden'
    )
)

# ============================================================
# INTERPRETATION PANEL
# ============================================================

interpretation_panel = HTML("""
<div style="
    width:325px;
    padding:8px 9px;
    border:1px solid #d7c0e4;
    font-family:Arial;
    font-size:13px;
    line-height:1.45;
    margin-top:3px;
    box-sizing:border-box;
">

<div style="
    font-size:16px;
    font-weight:bold;
    color:#6a1b9a;
    margin-bottom:5px;
">
Interpretation
</div>

<div style="margin-bottom:4px;">
The point <b>P</b> moves counterclockwise on the unit circle as the angle θ changes.
</div>

<div style="margin-bottom:4px;">
The projections of P on the coordinate axes give:
</div>

<div style="color:#1555cc; margin-left:10px;">
• cos(θ) → horizontal projection
</div>

<div style="color:#d7191c; margin-left:10px;">
• sin(θ) → vertical projection
</div>

<div style="margin-top:6px;">
The horizontal and vertical projections are therefore always between −1 and 1.
</div>

</div>
""")

# ============================================================
# RIGHT COLUMN
# ============================================================

right_column = VBox(
    [
        fig_circle.canvas,
        circle_values_panel,
        interpretation_panel
    ],
    layout=Layout(
        width='340px',
        min_width='340px',
        align_items='center',
        overflow='hidden'
    )
)

# ============================================================
# BOTTOM INFORMATION PANEL
# ============================================================

bottom_panel = HTML("""
<div style="
    width:710px;
    padding:7px 10px;
    border:1px solid #e2cf8c;
    background:#fffdf6;
    font-family:Arial;
    font-size:13px;
    line-height:1.45;
    box-sizing:border-box;
    margin-top:3px;
">

<b>Try changing the sliders A, f, φ and t and observe:</b>

<div style="
    display:grid;
    grid-template-columns:1fr 1fr;
    column-gap:24px;
    margin-top:4px;
">

<div>
• The sine and cosine waves update in real time.<br>
• The point P moves on the unit circle.
</div>

<div>
• The values of sin(θ) and cos(θ) remain between −1 and 1.<br>
• The identity <b>sin²(θ) + cos²(θ) = 1</b> is always satisfied.
</div>

</div>

</div>
""")

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    # --------------------------------------------------------
    # Current parameter values
    # --------------------------------------------------------

    A = amplitude_slider.value
    f = frequency_slider.value
    phi = phase_slider.value
    t_current = time_slider.value

    omega = 2.0 * np.pi * f

    theta = (
        omega * t_current
        + phi
    )

    theta_display = np.mod(
        theta,
        2.0 * np.pi
    )

    theta_deg = np.degrees(
        theta_display
    )

    sin_theta = np.sin(
        theta
    )

    cos_theta = np.cos(
        theta
    )

    sine_value = (
        A * sin_theta
    )

    cosine_value = (
        A * cos_theta
    )

    # --------------------------------------------------------
    # Signal curves
    # --------------------------------------------------------

    sine_wave = A * np.sin(
        omega * t_axis + phi
    )

    cosine_wave = A * np.cos(
        omega * t_axis + phi
    )

    line_sine.set_ydata(
        sine_wave
    )

    line_cosine.set_ydata(
        cosine_wave
    )

    # --------------------------------------------------------
    # Current-time marker
    # --------------------------------------------------------

    time_marker.set_xdata(
        [t_current, t_current]
    )

    point_sine.set_data(
        [t_current],
        [sine_value]
    )

    point_cosine.set_data(
        [t_current],
        [cosine_value]
    )

    ax_wave.set_ylim(
        -3.3,
        3.3
    )

    # --------------------------------------------------------
    # Unit circle
    # --------------------------------------------------------

    xp = cos_theta
    yp = sin_theta

    radius_line.set_data(
        [0.0, xp],
        [0.0, yp]
    )

    circle_point.set_data(
        [xp],
        [yp]
    )

    cos_projection.set_data(
        [0.0, xp],
        [0.0, 0.0]
    )

    sin_projection.set_data(
        [xp, xp],
        [0.0, yp]
    )

    # --------------------------------------------------------
    # Angle arc
    # --------------------------------------------------------

    arc_angles = np.linspace(
        0.0,
        theta_display,
        150
    )

    arc_radius = 0.28

    angle_arc.set_data(
        arc_radius * np.cos(arc_angles),
        arc_radius * np.sin(arc_angles)
    )

    # --------------------------------------------------------
    # Text positions
    # --------------------------------------------------------

    point_text.set_position(
        (
            xp + 0.07,
            yp + 0.07
        )
    )

    theta_label_angle = (
        theta_display / 2.0
    )

    theta_text.set_position(
        (
            0.36 * np.cos(theta_label_angle),
            0.36 * np.sin(theta_label_angle)
        )
    )

    # --------------------------------------------------------
    # Slider current values
    # --------------------------------------------------------

    amplitude_value.value = (
        f'<div class="trig-value">{A:.2f}</div>'
    )

    frequency_value.value = (
        f'<div class="trig-value">{f:.2f}</div>'
    )

    phase_value.value = (
        f'<div class="trig-value">{phi:.2f}</div>'
    )

    time_value.value = (
        f'<div class="trig-value">{t_current:.2f}</div>'
    )

    # --------------------------------------------------------
    # Current values panel
    # --------------------------------------------------------

    visibility = (
        'block'
        if show_values_checkbox.value
        else 'none'
    )

    current_values.value = f"""
    <div style="
        display:{visibility};
        font-family:Arial;
        font-size:14px;
        line-height:1.58;
    ">

    <div style="color:#16802b;">
    <b>t = {t_current:.2f} s</b>
    </div>

    <div>
    θ = ωt + φ =
    <b>{theta_display:.3f} rad</b>
    &nbsp; ({theta_deg:.1f}°)
    </div>

    <div style="color:#d7191c;">
    <b>sin(θ) = {sin_theta:.4f}</b>
    </div>

    <div style="color:#1555cc;">
    <b>cos(θ) = {cos_theta:.4f}</b>
    </div>

    <div style="margin-top:3px; color:#d7191c;">
    <b>A sin(θ) = {sine_value:.4f}</b>
    </div>

    <div style="color:#1555cc;">
    <b>A cos(θ) = {cosine_value:.4f}</b>
    </div>

    <hr style="margin:7px 0px;">

    <div style="text-align:center;">
    sin²(θ) + cos²(θ)
    =
    <b>{sin_theta**2 + cos_theta**2:.6f}</b>
    </div>

    </div>
    """

    # --------------------------------------------------------
    # Equations panel
    # --------------------------------------------------------

    period = 1.0 / f

    equations_html.value = f"""
    <div style="
        font-family:Arial;
        font-size:13px;
        line-height:1.50;
    ">

    <div style="
        display:grid;
        grid-template-columns:1fr 1fr;
        column-gap:20px;
    ">

    <div style="color:#d7191c;">
    <b>y₁(t) = A sin(ωt + φ)</b><br>
    = {A:.2f} sin(2π · {f:.2f} t + {phi:.2f})
    </div>

    <div style="color:#1555cc;">
    <b>y₂(t) = A cos(ωt + φ)</b><br>
    = {A:.2f} cos(2π · {f:.2f} t + {phi:.2f})
    </div>

    </div>

    <div style="
        margin-top:6px;
        padding:4px 7px;
        background:#fff9e8;
        border:1px solid #e3d39c;
        text-align:center;
    ">
    ω = 2πf = <b>{omega:.4f} rad/s</b>
    &nbsp;&nbsp;&nbsp;
    T = 1/f = <b>{period:.4f} s</b>
    </div>

    </div>
    """

    # --------------------------------------------------------
    # Unit-circle values
    # --------------------------------------------------------

    circle_values_html.value = f"""
    <div style="
        font-family:Arial;
        font-size:13px;
        line-height:1.55;
        text-align:center;
    ">

    <div style="color:#d7191c;">
    <b>sin(θ) = y<sub>P</sub> = {sin_theta:.4f}</b>
    </div>

    <div style="color:#1555cc;">
    <b>cos(θ) = x<sub>P</sub> = {cos_theta:.4f}</b>
    </div>

    <div style="
        margin-top:6px;
        border:1px solid #d7c0e4;
        padding:6px;
        color:#6a1b9a;
    ">

    θ = ωt + φ<br>

    = {omega:.4f} · {t_current:.2f}
      + {phi:.2f}<br>

    = <b>{theta_display:.4f} rad
    ({theta_deg:.1f}°)</b>

    </div>

    </div>
    """

    # --------------------------------------------------------
    # REDRAW ONLY
    # --------------------------------------------------------

    fig_wave.canvas.draw_idle()
    fig_circle.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

amplitude_slider.observe(
    update_notebook,
    names='value'
)

frequency_slider.observe(
    update_notebook,
    names='value'
)

phase_slider.observe(
    update_notebook,
    names='value'
)

time_slider.observe(
    update_notebook,
    names='value'
)

show_values_checkbox.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# LEFT + CENTRAL UPPER BLOCK
# ============================================================

left_center_top = HBox(
    [
        left_column,
        wave_box
    ],
    layout=Layout(
        width='925px',
        min_width='925px',
        align_items='flex-start',
        gap='10px',
        overflow='hidden'
    )
)

# ============================================================
# LEFT + CENTRAL BLOCK
# ============================================================

left_center_block = VBox(
    [
        left_center_top,
        bottom_panel
    ],
    layout=Layout(
        width='925px',
        min_width='925px',
        gap='2px',
        overflow='hidden'
    )
)

# ============================================================
# COMPLETE DASHBOARD
#
# Total internal width is now roughly:
#
# 925 + 10 + 340 = 1275 px
#
# This is much smaller than the previous 1600 px layout.
# ============================================================

dashboard = HBox(
    [
        left_center_block,
        right_column
    ],
    layout=Layout(
        width='1275px',
        max_width='100%',
        align_items='flex-start',
        gap='10px',
        overflow='hidden'
    )
)

# ============================================================
# COMPLETE NOTEBOOK
#
# width = 100% is important for Binder / slide embedding.
# max_width prevents creation of a huge output canvas.
# ============================================================

main_layout = VBox(
    [
        dashboard
    ],
    layout=Layout(
        width='100%',
        max_width='1275px',
        overflow='hidden'
    )
)

display(main_layout)
```
