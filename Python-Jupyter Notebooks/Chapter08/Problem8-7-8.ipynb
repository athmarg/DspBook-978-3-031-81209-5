```python
import sympy as sp

# Enable pretty printing
sp.init_printing(use_unicode=True)

# Define symbols
z = sp.Symbol('z', complex=True)
n = sp.Symbol('n', integer=True)

# 1. Define the function X(z)
X_z = 1 / (1 + 3 * z**(-1) + 2 * z**(-2))

print("1. The function X(z) is:")
display(X_z)

# 2. Partial Fraction Expansion
# To do this symbolically with z, we temporarily substitute z^(-1) with y
y = sp.Symbol('y')
X_y = 1 / (1 + 3*y + 2*y**2)

# Compute partial fractions with respect to y
partial_frac_y = sp.apart(X_y, y)

print("\n2. Partial fraction expansion (in terms of y = z^-1):")
display(partial_frac_y)

# 3. Return to the original variable z (substitute y = z^-1)
partial_frac_z = partial_frac_y.subs(y, 1/z)

print("\n3. Partial fraction expansion in terms of z:")
display(partial_frac_z)

# 4. Computation of the Inverse Z-Transform
# We know that the inverse of 1 / (1 - a * z^-1) is a^n * u[n]
# Therefore:
# 2 / (1 + 2*z^-1) = 2 / (1 - (-2)*z^-1) -> 2 * (-2)^n * u[n]
# -1 / (1 + z^-1)   = -1 / (1 - (-1)*z^-1) -> -1 * (-1)^n * u[n]

u_n = sp.Heaviside(n) # Step function u[n]

# Compute the final signal x[n]
x_n = 2 * ((-2)**n) * u_n - ((-1)**n) * u_n

print("\n4. The final inverse Z-transform x[n] is:")
display(x_n)

# Simplified form
x_n_simplified = sp.simplify(x_n)
print("In simplified form:")
display(x_n_simplified)
```
