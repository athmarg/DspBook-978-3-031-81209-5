```python
import sympy as sp
from IPython.display import display

# ==============================================================================
# 1. SymPy Settings and Symbol Definitions
# ==============================================================================

sp.init_printing(use_unicode=True)

z = sp.Symbol('z', complex=True)
n = sp.Symbol('n', integer=True, nonnegative=True)

print("=== METHOD: Complex Integration Method (Residues) with exact Rationals ===")

# 1. Define raw X(z) and CONVERT to exact rational fractions to prevent limit(0) bugs
X_z_raw = (z - 1.5) / (z**2 - 0.7 * z + 0.1)
X_z = sp.nsimplify(X_z_raw)

print("\n1. Function X(z) strictly as exact fractions:")
display(X_z)

# 2. Base function for residues (without z^n to avoid symbolic limit confusion)
Y_z_base = X_z / z

print("\n2. Base function Y(z)_base = X(z) / z:")
display(Y_z_base)

# Find EXACT poles
num_y, den_y = sp.fraction(sp.cancel(Y_z_base))
poles = sp.solve(den_y, z)

print("\n3. Detected exact poles (including zero):")
display(poles)

# Calculate sum of residues
res_sum = 0
delta = sp.KroneckerDelta(n, 0) # Dirac Delta for n=0
u = sp.Heaviside(n)             # Step function for n>=0

print("\n4. Calculating individual residues:")
for pole in poles:
    # We calculate the algebraic coefficient explicitly
    coeff = sp.simplify(sp.limit((z - pole) * Y_z_base, z, pole))
    
    if pole == 0:
        # Pole at zero corresponds to the delta function term A * \delta[n]
        term = coeff * delta
        print(f" -> Pole at z = {pole}: Coefficient = {coeff} (Forms {coeff}*δ[n])")
    else:
        # Non-zero poles correspond to terms A * (p)^n * u[n]
        term = coeff * (pole**n) * u
        print(f" -> Pole at z = {pole}: Coefficient = {coeff.evalf(5)} (Forms {coeff} * ({pole})^n)")
        
    res_sum += term

# Final Inverse Z-Transform expression
x_n = sp.simplify(res_sum)

print("\n5. Final inverse Z-transform signal x[n]:")
display(x_n)
```
