```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML

# Disable auto-scroll in notebook output
display(HTML("<style>.output_scroll { height: unset !important; }</style>"))

def polyphase_frequency_domain_demo():
    # 1. Define an example low-pass impulse response h[n] of length 12
    h = np.array([0.02, 0.05, 0.12, 0.22, 0.30, 0.38, 0.38, 0.30, 0.22, 0.12, 0.05, 0.02])
    M = 3  # Polyphase factor
    
    # 2. Extract Polyphase Components e_k[n]
    e0 = h[0::3]
    e1 = h[1::3]
    e2 = h[2::3]
    
    # 3. Frequency response calculation using DTFT
    N_fft = 2048
    omega = np.linspace(-np.pi, np.pi, N_fft)
    
    # Frequency response of the original filter H(e^{j\omega})
    H_omega = np.zeros(N_fft, dtype=complex)
    for n, val in enumerate(h):
        H_omega += val * np.exp(-1j * omega * n)
        
    # Frequency responses of the polyphase components E_k(e^{j\omega M})
    E0_omega = np.zeros(N_fft, dtype=complex)
    for n, val in enumerate(e0):
        E0_omega += val * np.exp(-1j * (omega * M) * n)
        
    E1_omega = np.zeros(N_fft, dtype=complex)
    for n, val in enumerate(e1):
        E1_omega += val * np.exp(-1j * (omega * M) * n)
        
    E2_omega = np.zeros(N_fft, dtype=complex)
    for n, val in enumerate(e2):
        E2_omega += val * np.exp(-1j * (omega * M) * n)
        
    # Reconstruction verification in frequency domain:
    H_reconstructed = E0_omega + np.exp(-1j * omega * 1) * E1_omega + np.exp(-1j * omega * 2) * E2_omega

    # 4. Plotting Magnitude and Phase Responses
    fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
    
    # --- Magnitude Response ---
    axes[0].plot(omega / np.pi, np.abs(H_omega), color='black', lw=2.5, label=r'Original Filter $|\mathcal{H}(e^{j\omega})|$')
    axes[0].plot(omega / np.pi, np.abs(E0_omega), color='tab:orange', lw=1.5, linestyle='--', label=r'Polyphase Component $|E_0(e^{j\omega M})|$')
    axes[0].plot(omega / np.pi, np.abs(E1_omega), color='tab:green', lw=1.5, linestyle='--', label=r'Polyphase Component $|E_1(e^{j\omega M})|$')
    axes[0].plot(omega / np.pi, np.abs(E2_omega), color='tab:purple', lw=1.5, linestyle='--', label=r'Polyphase Component $|E_2(e^{j\omega M})|$')
    axes[0].plot(omega / np.pi, np.abs(H_reconstructed), color='red', lw=1.5, linestyle=':', label=r'Reconstructed Sum')
    
    axes[0].set_title(r'Polyphase Analysis: Magnitude Frequency Response', fontsize=11, fontweight='bold')
    axes[0].set_ylabel('Magnitude', fontsize=10)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    # Legend placed outside to the right
    axes[0].legend(loc='center left', bbox_to_anchor=(1.02, 0.5), frameon=True, fontsize=9)
    axes[0].set_xlim(-1, 1)
    
    # --- Phase Response ---
    axes[1].plot(omega / np.pi, np.angle(H_omega), color='black', lw=2.5, label=r'Original Filter Phase $\angle \mathcal{H}(e^{j\omega})$')
    axes[1].plot(omega / np.pi, np.angle(E0_omega), color='tab:orange', lw=1.5, linestyle='--', label=r'Phase $\angle E_0(e^{j\omega M})$')
    axes[1].plot(omega / np.pi, np.angle(E1_omega), color='tab:green', lw=1.5, linestyle='--', label=r'Phase $\angle E_1(e^{j\omega M})$')
    axes[1].plot(omega / np.pi, np.angle(E2_omega), color='tab:purple', lw=1.5, linestyle='--', label=r'Phase $\angle E_2(e^{j\omega M})$')
    
    axes[1].set_title(r'Polyphase Analysis: Phase Frequency Response', fontsize=11, fontweight='bold')
    axes[1].set_xlabel(r'Normalized Frequency ($\omega / \pi$)', fontsize=10)
    axes[1].set_ylabel('Phase (radians)', fontsize=10)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    # Legend placed outside to the right
    axes[1].legend(loc='center left', bbox_to_anchor=(1.02, 0.5), frameon=True, fontsize=9)
    axes[1].set_xlim(-1, 1)
    
    plt.tight_layout()
    plt.show()
    
    # Conclusion Box
    display(HTML(r"""
    <div style="border: 1px solid #ccc; padding: 12px; border-radius: 5px; background-color: #f9f9f9; font-family: sans-serif; font-size: 14px;">
        <strong>Conclusion:</strong> 
        This frequency-domain visualization illustrates how the overall magnitude and phase response of the original filter $\mathcal{H}(e^{j\omega})$ (black curve) is formed through the weighted superposition of the individual polyphase component responses $E_k(e^{j\omega M})$ combined with their respective delay terms ($e^{-j\omega k}$), confirming the theoretical foundations of polyphase filter structures.
    </div>
    """))

# Run the demonstration function
polyphase_frequency_domain_demo()
```
