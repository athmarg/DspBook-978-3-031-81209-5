```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown, clear_output
from ipywidgets import Output, VBox, Layout

text_out = Output(layout=Layout(width='99%', overflow='hidden'))
fig_out = Output(layout=Layout(width='99%', overflow='hidden'))

def update_visualization():
    with text_out:
        clear_output(wait=True)
        display(Markdown("### Natural Response Analysis - Exercise 1"))
        display(Markdown(r"**Difference Equation:** $y[n] - 3y[n-1] - 4y[n-2] = 0$"))
        display(Markdown(r"**Initial Conditions:** $y[-1] = 5,\ y[-2] = 0$"))
        display(Markdown(r"**Constants:** $C_1 = 16,\ C_2 = -1$"))
        display(Markdown(r"**Analytical Natural Response:** $y_{\text{zi}}[n] = 16(4)^n + (-1)(-1)^n = (4)^{n+2} + (-1)^{n+1}$"))

    with fig_out:
        clear_output(wait=True)
        
        n_vals = np.arange(0, 11)
        y_vals = 16 * (4.0 ** n_vals) - ((-1.0) ** n_vals)

        fig, ax = plt.subplots(figsize=(9.5, 4.5))
        ax.stem(n_vals, y_vals, linefmt='b-', markerfmt='bo', basefmt='k-', label=r'$y_{\text{zi}}[n]$')
        ax.set_title(r'Natural Response $y_{\text{zi}}[n]$', fontsize=11, fontweight='bold', color='darkblue')
        ax.set_xlabel('Time index n', fontsize=10)
        ax.set_ylabel('Amplitude', fontsize=10)
        ax.grid(True, linestyle='--', alpha=0.6)
        ax.legend(loc='upper left')

        plt.tight_layout()
        plt.show()

update_visualization()
ui = VBox([text_out, fig_out], layout=Layout(align_items='flex-start', width='99%', overflow='hidden'))
display(ui)
```
