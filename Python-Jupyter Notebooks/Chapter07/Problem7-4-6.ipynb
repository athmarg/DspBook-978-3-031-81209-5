```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown
import sympy as sp

# Define parameters for the sinusoidal input signal x[n] = A * cos(omega_0 * n + phi)
A = 2.5
omega_0 = np.pi / 4.0  # Input frequency (rad/sample)
phi = np.pi / 6.0      # Phase offset (rad)

display(Markdown("### LTI System Frequency Response Analysis"))
display(Markdown(r"**Input Signal:** $x[n] = " + f"{A}" + r"\cos(" + f"{omega_0:.3f}" + r"n + " + f"{phi:.3f}" + r")$"))

# Define system impulse response h[n]
n_vals = np.arange(0, 41)
h = (0.5) ** n_vals

# Frequency response evaluation over omega in [-pi, pi]
omega_vals = np.linspace(-np.pi, np.pi, 1000)
H_omega = np.array([np.sum(h * np.exp(-1j * w * np.arange(len(h)))) for w in omega_vals])

# Evaluate magnitude and phase at the specific input frequency omega_0
H_val = np.sum(h * np.exp(-1j * omega_0 * np.arange(len(h))))
mag_H = np.abs(H_val)
phase_H = np.angle(H_val)

display(Markdown(r"**Calculated Values at $\omega_0$:** $|\mathcal{H}(e^{j\omega_0})| = " + f"{mag_H:.3f}" + r"$, $\angle\mathcal{H}(e^{j\omega_0}) = " + f"{phase_H:.3f}" + r"\text{ rad}$"))

# Analytical Symbolic Output for Output Signal y[n]
n = sp.symbols('n', integer=True)
y_symbolic = A * sp.Float(mag_H, 3) * sp.cos(sp.Float(omega_0, 3) * n + sp.Float(phi, 3) + sp.Float(phase_H, 3))

display(Markdown(r"**System Output Expression $y[n]$:**"))
display(sp.Eq(sp.Symbol('y[n]'), y_symbolic))

# Numerical evaluation of input and output signals for plotting
x = A * np.cos(omega_0 * n_vals + phi)
y = A * mag_H * np.cos(omega_0 * n_vals + phi + phase_H)

# Plotting: Signals and Frequency Response (Magnitude & Phase with omega_0 markers)
fig, axes = plt.subplots(3, 1, figsize=(10, 10))

# 1. Input and Output Signals
axes[0].stem(n_vals, x, linefmt='b-', markerfmt='bo', basefmt='k-', label='Input $x[n]$')
axes[0].stem(n_vals, y, linefmt='r-', markerfmt='ro', basefmt='k-', label='Output $y[n]$')
axes[0].set_title(r'Input and Output Signals', fontsize=9.5, fontweight='bold', color='darkblue')
axes[0].set_ylabel('Amplitude', fontsize=8.5)
axes[0].grid(True, linestyle='--', alpha=0.6)
axes[0].legend(loc='upper right')

# 2. Magnitude Response with vertical line at omega_0
axes[1].plot(omega_vals, np.abs(H_omega), 'b-', linewidth=1.5, label=r'$|\mathcal{H}(e^{j\omega})|$')
axes[1].axvline(x=omega_0, color='g', linestyle='--', label=r'Input Frequency $\omega_0$')
axes[1].scatter([omega_0], [mag_H], color='g', zorder=5)
axes[1].text(omega_0 + 0.1, mag_H + 0.35, r'$|\mathcal{H}(e^{j\omega_0})| \approx ' + f'{mag_H:.3f}$', color='g', fontweight='bold', va='bottom')
axes[1].set_title(r'Magnitude Response $|\mathcal{H}(e^{j\omega})|$', fontsize=9.5, fontweight='bold', color='darkblue')
axes[1].set_ylabel('Magnitude', fontsize=8.5)
axes[1].grid(True, linestyle='--', alpha=0.6)
axes[1].legend(loc='upper right')

# 3. Phase Response with vertical line and marker at omega_0
axes[2].plot(omega_vals, np.angle(H_omega), 'r-', linewidth=1.5, label=r'$\angle\mathcal{H}(e^{j\omega})$')
axes[2].axvline(x=omega_0, color='g', linestyle='--', label=r'Input Frequency $\omega_0$')
axes[2].scatter([omega_0], [phase_H], color='g', zorder=5)
axes[2].text(omega_0 + 0.1, phase_H + 0.35, r'$\angle\mathcal{H}(e^{j\omega_0}) \approx ' + f'{phase_H:.3f}\\text{{ rad}}$', color='g', fontweight='bold', va='bottom')
axes[2].set_title(r'Phase Response $\angle\mathcal{H}(e^{j\omega})$', fontsize=9.5, fontweight='bold', color='darkred')
axes[2].set_xlabel(r'Frequency $\omega$ (rad/sample)', fontsize=8.5)
axes[2].set_ylabel('Phase (rad)', fontsize=8.5)
axes[2].grid(True, linestyle='--', alpha=0.6)
axes[2].legend(loc='upper right')

plt.tight_layout()
plt.show()
```
