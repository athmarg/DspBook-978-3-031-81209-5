```python
import numpy as np

for N in range(1, 9):
    n = np.arange(N)
    k = n.reshape((N, 1))
    W_N = np.exp(-2j * np.pi * k * n / N)
    
    print(f"--- W_{N} (N = {N}) ---")
    # Στρογγυλοποίηση στα 3 δεκαδικά ψηφία για καθαρή εκτύπωση
    rounded_W = np.round(W_N.real, 3) + 1j * np.round(W_N.imag, 3)
    print(rounded_W)
    print("\n")
```
