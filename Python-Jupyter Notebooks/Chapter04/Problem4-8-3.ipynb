```python
# Notebook: Solving Differential Equations via Laplace Transform
# Example: x''(t) + 5x'(t) + 6x(t) = e^{-t}, with x(0) = 2, x'(0) = 1

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- 1. SYMPY: Differential Equation Setup & Laplace Solution ---
s = sp.Symbol('s')
t = sp.Symbol('t', real=True, positive=True)

# Define X(s) as derived from the transformed differential equation
numerator = 2*s**2 + 13*s + 12
denominator = (s + 1) * (s**2 + 5*s + 6) # or (s+1)*(s+2)*(s+3)
X_s = numerator / denominator

print("1. Transformed Solution X(s):")
display(X_s)

# Perform Partial Fraction Expansion to verify coefficients
X_s_pfe = sp.apart(X_s, s)
print("Partial Fraction Expansion of X(s):")
display(X_s_pfe)

# Compute Inverse Laplace Transform to get x(t)
x_t = sp.inverse_laplace_transform(X_s, s, t)
print("\n2. Computed Solution x(t):")
display(x_t)


# --- 2. NUMPY & MATPLOTLIB: Time-domain response visualization ---
x_t_numeric = sp.lambdify(t, x_t, 'numpy')

t_vals = np.linspace(0, 5, 400)
x_vals = x_t_numeric(t_vals)

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(t_vals, x_vals, color='crimson', lw=2.5, label=r'$x(t) = \left(\frac{1}{2}e^{-t} + 6e^{-2t} - \frac{9}{2}e^{-3t}\right)u(t)$')

ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlabel(r'Time $t$', fontsize=11)
ax.set_ylabel(r'$x(t)$', fontsize=11)
ax.set_title(r'Differential Equation Response via Laplace Transform', fontsize=12)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='best', frameon=True)

plt.tight_layout()
plt.show()
```
