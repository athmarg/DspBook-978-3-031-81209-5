```python
# Notebook: Analytical Laplace Transform, ROC, and Visualization for x(t) = sin(\omega_0 t) u(t)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

# 1. Define symbolic variables and time-domain signal
t = sp.Symbol('t', real=True)
sigma, omega, omega_0 = sp.symbols('sigma omega omega_0', real=True)
s = sigma + sp.I * omega  # Complex variable s = \sigma + j\omega

# Define signal x(t) = sin(\omega_0 t) * u(t)
x_t = sp.sin(omega_0 * t) * sp.Heaviside(t)

print("1. Time-domain signal x(t):")
display(x_t)

# 2. Compute Laplace Transform with convergence conditions
result, var, cond = sp.laplace_transform(x_t, t, s, noconds=False)

print("\n2. Analytical Laplace Transform X(s):")
display(result)

print("\n3. Region of Convergence (ROC) condition:")
display(cond)

# ΔΙΟΡΘΩΣΗ: Προσθήκη 'r' μπροστά από το string (raw string) για να αγνοηθεί το \o
print(r"\n4. Numerical expression example (for \omega_0 = 2):")
omega_0_val = 2.0
x_t_num = sp.sin(omega_0_val * t) * sp.Heaviside(t)
result_num = sp.laplace_transform(x_t_num, t, s, noconds=True)
display(result_num)

# 5. Visualizing the Region of Convergence (ROC) in the complex s-plane (for \omega_0 = 2)
print(r"\n5. Region of Convergence (ROC) Plot in the s-plane:")
fig, ax = plt.subplots(figsize=(7, 6))

sig_vals = np.linspace(-3, 3, 400)
om_vals = np.linspace(-5, 5, 400)
SIGMA, OMEGA = np.meshgrid(sig_vals, om_vals)

# ROC condition for sine: \sigma > 0
roc_mask = SIGMA > 0.0

ax.imshow(roc_mask, extent=(-3, 3, -5, 5), origin='lower', cmap='Greens', alpha=0.3, aspect='auto')
ax.axvline(x=0.0, color='red', linestyle='--', linewidth=2, label=r'Boundary $\sigma = 0$ ($j\omega$ axis)')

# Poles at s = +j\omega_0 and s = -j\omega_0 (for \omega_0 = 2)
ax.plot(0.0, 2.0, 'rx', markersize=10, markeredgewidth=2, label=r'Poles at $s = \pm j\omega_0$')
ax.plot(0.0, -2.0, 'rx', markersize=10, markeredgewidth=2)

ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlim(-3, 3)
ax.set_ylim(-5, 5)
ax.set_xlabel(r'Real part $\sigma = \text{Re}\{s\}$', fontsize=11)
ax.set_ylabel(r'Imaginary part $j\omega = \text{Im}\{s\}$', fontsize=11)
ax.set_title(r'Region of Convergence (ROC) for $X(s) = \frac{\omega_0}{s^2+\omega_0^2}$', fontsize=12)
ax.grid(True, linestyle=':', alpha=0.7)
ax.legend(loc='upper right', fontsize=10)

plt.tight_layout()
plt.show()
```
