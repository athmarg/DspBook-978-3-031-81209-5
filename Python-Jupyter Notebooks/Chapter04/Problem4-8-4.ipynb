```python
# Notebook: Solving Second-Order Differential Equation with Resonance via Laplace Transform
# Example: x''(t) + 4x(t) = sin(2t), with x(0) = 0, x'(0) = 0 -> x(t) = (1/8)*sin(2t) - (t/4)*cos(2t)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- 1. SYMPY: Differential Equation Setup & Laplace Solution ---
s = sp.Symbol('s')
t = sp.Symbol('t', real=True, positive=True)

# Define X(s) = 2 / (s^2 + 4)^2
numerator = 2
denominator = (s**2 + 4)**2
X_s = numerator / denominator

print("1. Transformed Solution X(s):")
display(X_s)

# Compute Inverse Laplace Transform to get x(t)
x_t = sp.inverse_laplace_transform(X_s, s, t)
print("\n2. Computed Solution x(t):")
display(x_t)


# --- 2. NUMPY & MATPLOTLIB: Time-domain response visualization ---
x_t_numeric = sp.lambdify(t, x_t, 'numpy')

t_vals = np.linspace(0, 15, 600)
x_vals = x_t_numeric(t_vals)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(t_vals, x_vals, color='crimson', lw=2, label=r'$x(t) = \frac{1}{8}\sin(2t) - \frac{t}{4}\cos(2t)$')

ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlabel(r'Time $t$', fontsize=11)
ax.set_ylabel(r'$x(t)$', fontsize=11)
ax.set_title(r'Resonant Response of Differential Equation to $\sin(2t)$', fontsize=12)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='upper left', frameon=True)

plt.tight_layout()
plt.show()
```
