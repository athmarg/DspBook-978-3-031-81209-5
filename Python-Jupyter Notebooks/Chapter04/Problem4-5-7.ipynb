```python
# Notebook: Laplace Transform using Shift and Frequency Differentiation Properties
# Example: x(t) = t * e^{\alpha t} * cos(\beta t)

import sympy as sp
from IPython.display import display

# 1. Define symbolic variables and parameters
t = sp.Symbol('t', real=True)
s = sp.Symbol('s')
alpha = sp.Symbol('alpha', real=True)
beta = sp.Symbol('beta', positive=True, real=True)

# 2. Base transform: L{cos(\beta t)}
X_cos = s / (s**2 + beta**2)

print("1. Standard transform L{cos(\beta t)}:")
display(X_cos)

# 3. Apply differentiation property for L{t * cos(\beta t)} = -d/ds { X(s) }
# (Note: sp.diff gives d/ds, and we negate it)
X_t_cos = -sp.diff(X_cos, s)

print("\n2. After applying differentiation property for L{t * cos(\beta t)}:")
display(X_t_cos)

# Simplify the derivative expression algebraically
X_t_cos_simplified = sp.simplify(X_t_cos)
display(X_t_cos_simplified)

# 4. Apply frequency shifting property: L{e^{\alpha t} * x(t)} -> substitute s with (s - alpha)
final_result = X_t_cos_simplified.subs(s, s - alpha)

print("\n3. Final result after frequency shift (s -> s - alpha):")
display(final_result)

# Fully expand/simplify to match the final fraction of the slide
final_simplified = sp.simplify(final_result)
print("\n4. Fully simplified final expression:")
display(final_simplified)
```
