```python
# Notebook: Inverse Laplace Transform via Complex Integration (Bromwich Contour)
# Example: X(s) = exp(-\alpha s) / s -> x(t) = u(t - \alpha) (Heaviside Step Function)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- 1. SYMPY: Symbolic calculation & Residues ---
s = sp.Symbol('s')
t = sp.Symbol('t', real=True)
alpha = sp.Symbol('alpha', positive=True, real=True)

# Define the transform X(s) = exp(-alpha*s) / s
X_s = sp.exp(-alpha * s) / s

print("1. Given Transform X(s):")
display(X_s)

# Function inside the Bromwich integral: X(s) * exp(s*t) = exp((t - alpha)*s) / s
integrand = X_s * sp.exp(s * t)
print("Integrand X(s) * exp(s*t):")
display(integrand)

# Find the pole at s = 0
poles = sp.solve(s, s)
print("Pole of the function:", poles)

# Compute residue at s = 0 via SymPy
res_0 = sp.limit(integrand * s, s, 0)
print("\nResidue at s = 0:")
display(res_0)

# Inverse Laplace transform using SymPy's built-in engine
x_t_symbolic = sp.inverse_laplace_transform(X_s, s, t)
print("\n2. Computed Inverse Transform x(t) via SymPy:")
display(x_t_symbolic)


# --- 2. MATPLOTLIB: Visualization of the Bromwich Contours (Figure 0406) ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 7))

delta = 2.0  # Abscissa of convergence Re(s) = delta
R = 5.0      # Radius of semicircle

# --- Left Image: Case t > alpha (Closed to the left enclosing the pole at s = 0) ---
# 1. Pole at s = 0
ax1.scatter([0], [0], color='red', s=120, zorder=5, label=r'Pole ($s = 0$)')

# 2. Bromwich path and left semicircle
y_vals = np.linspace(-R, R, 200)
ax1.plot([delta]*len(y_vals), y_vals, color='blue', lw=2.5, label=r'Bromwich Path ($\delta + j\omega$)')

theta_left = np.linspace(np.pi/2, 3*np.pi/2, 150)
semi_re_1 = R * np.cos(theta_left)
semi_im_1 = R * np.sin(theta_left)
ax1.plot(semi_re_1, semi_im_1, color='blue', ls='--', lw=1.5, label=r'Left Semicircle $C_R$ ($t > \alpha$)')

ax1.plot([delta, 0], [R, R], color='blue', ls='--', lw=1.5)
ax1.plot([delta, 0], [-R, -R], color='blue', ls='--', lw=1.5)

ax1.annotate('', xy=(delta, R/2), xytext=(delta, R/2 - 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))
ax1.annotate('', xy=(delta, -R/2), xytext=(delta, -R/2 + 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))

ax1.axhline(0, color='black', linewidth=1)
ax1.axvline(0, color='black', linewidth=1)
ax1.set_xlim(-6, 4)
ax1.set_ylim(-6, 6)
ax1.set_xlabel(r'Real part ($\sigma$)', fontsize=12)
ax1.set_ylabel(r'Imaginary part ($j\omega$)', fontsize=12)
ax1.set_title(r'Case $t > \alpha$ (Encloses Pole $s = 0$)', fontsize=13)
ax1.grid(True, linestyle=':', alpha=0.6)

# Legend placed below the first subplot
ax1.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=True)


# --- Right Image: Case t < alpha (Closed to the right, no poles enclosed) ---
# 1. Pole at s = 0
ax2.scatter([0], [0], color='red', s=120, zorder=5, label=r'Pole ($s = 0$)')

# 2. Bromwich path and right semicircle
ax2.plot([delta]*len(y_vals), y_vals, color='blue', lw=2.5, label=r'Bromwich Path ($\delta + j\omega$)')

theta_right = np.linspace(-np.pi/2, np.pi/2, 150)
semi_re_2 = R * np.cos(theta_right)
semi_im_2 = R * np.sin(theta_right)
ax2.plot(semi_re_2, semi_im_2, color='blue', ls='--', lw=1.5, label=r'Right Semicircle $C_R$ ($t < \alpha$)')

ax2.plot([delta, 0], [R, R], color='blue', ls='--', lw=1.5)
ax2.plot([delta, 0], [-R, -R], color='blue', ls='--', lw=1.5)

ax2.annotate('', xy=(delta, R/2), xytext=(delta, R/2 - 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))
ax2.annotate('', xy=(delta, -R/2), xytext=(delta, -R/2 + 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))

ax2.axhline(0, color='black', linewidth=1)
ax2.axvline(0, color='black', linewidth=1)
ax2.set_xlim(-4, 6)
ax2.set_ylim(-6, 6)
ax2.set_xlabel(r'Real part ($\sigma$)', fontsize=12)
ax2.set_ylabel(r'Imaginary part ($j\omega$)', fontsize=12)
ax2.set_title(r'Case $t < \alpha$ (No Poles Enclosed)', fontsize=13)
ax2.grid(True, linestyle=':', alpha=0.6)

# Legend placed below the second subplot
ax2.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=True)

# Main title and layout adjustments
plt.suptitle(r'Contour Integration for $X(s) = \frac{e^{-\alpha s}}{s}$', fontsize=15, y=0.98)

# Adjust layout to make room for the legends below the plots
plt.tight_layout(rect=[0, 0.05, 1, 0.95])
plt.show()
```
