```python
# Notebook: Laplace Transform using Frequency Differentiation Property
# Example: x(t) = t^2 * cos(\omega_0 t)

import sympy as sp
from IPython.display import display

# 1. Define symbolic variables and parameters
t = sp.Symbol('t', real=True)
s = sp.Symbol('s')
omega_0 = sp.Symbol('omega', positive=True, real=True)

# 2. Base transform: L{cos(\omega_0 t)}
X_cos = s / (s**2 + omega_0**2)

print(r"1. Standard transform L{cos(\omega_0 t)}:")
display(X_cos)

# 3. Apply first differentiation in s-domain: L{-t * cos(\omega_0 t)} = d/ds { X(s) }
step_1 = sp.diff(X_cos, s)

print(r"2. After first differentiation (corresponding to L{-t * cos(\omega_0 t)}):")
display(step_1)

# Simplify the first derivative algebraically to match the slide
step_1_simplified = sp.simplify(step_1)
display(step_1_simplified)

# 4. Apply second differentiation: L{t^2 * cos(\omega_0 t)} = d^2/ds^2 { X(s) }
step_2 = sp.diff(X_cos, s, 2)

print(r"3. Final result after second differentiation L{t^2 * cos(\omega_0 t)}:")
display(step_2)

# Simplify completely to match the final fraction of the slide
final_result = sp.simplify(step_2)
print(r"4. Fully simplified final expression:")
display(final_result)
```
