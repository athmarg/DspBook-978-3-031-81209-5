```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, RadioButtons, HBox, Layout, VBox, HTML
from IPython.display import display

def plot_filter_response(filter_type='Low-pass', Q=5.0):
    fig, ax = plt.subplots(figsize=(9, 6))

    freq = np.logspace(-1, 1, 1000)
    omega = freq

    if filter_type == 'Low-pass':
        H_mag = 1.0 / np.sqrt((1.0 - omega**2)**2 + (omega / Q)**2)
        title_str = "Low-pass filter"
        y_min, y_max = -50, 30

    elif filter_type == 'High-pass':
        H_mag = omega**2 / np.sqrt((1.0 - omega**2)**2 + (omega / Q)**2)
        title_str = "High-pass filter"
        y_min, y_max = -50, 30

    elif filter_type == 'Band-pass':
        H_mag = (omega / Q) / np.sqrt((1.0 - omega**2)**2 + (omega / Q)**2)
        title_str = "Band-pass filter"
        y_min, y_max = -70, 10

    elif filter_type == 'Band-stop':
        H_mag = np.abs(1.0 - omega**2) / np.sqrt((1.0 - omega**2)**2 + (omega / Q)**2)
        title_str = "Band-stop filter"
        y_min, y_max = -50, 5

    gain_db = 20 * np.log10(np.clip(H_mag, 1e-5, 100))

    ax.semilogx(freq, gain_db, 'r-', linewidth=1.5)
    ax.set_xlim(0.1, 10)
    ax.set_ylim(y_min, y_max)
    ax.set_xlabel('Frequency (Hz)', fontsize=12)
    ax.set_ylabel('Gain (dB)', fontsize=12)
    ax.grid(True, which="both", ls=":", color='gray', alpha=0.7)
    ax.axvline(1.0, color='gray', linestyle='-', linewidth=0.8)
    ax.axhline(0.0, color='gray', linestyle='-', linewidth=0.8)
    ax.set_xticks([0.1, 1, 10])
    ax.set_xticklabels(['0.1', '1', '10'])
    ax.set_title(f"Gain response of {title_str.lower()} with Q = {Q:.2f}", fontsize=13, pad=12)

    plt.tight_layout()
    plt.show()

filter_title = HTML(value="<b>Filter Type:</b>", layout=Layout(margin='0 0 4px 0'))

filter_radio = RadioButtons(options=['Low-pass', 'High-pass', 'Band-pass', 'Band-stop'], value='Low-pass', description='', disabled=False, layout=Layout(width='220px'))

q_title = HTML(value="<b>Q Factor:</b>", layout=Layout(margin='12px 0 4px 0'))

q_slider = FloatSlider(min=0.1, max=20.0, step=0.1, value=5.0, description='', readout=True, readout_format='.2f', style={'description_width': '0px'}, layout=Layout(width='280px'))

interactive_plot = interactive(plot_filter_response, filter_type=filter_radio, Q=q_slider)

interactive_plot.children[0].description = ''
interactive_plot.children[1].description = ''

controls_box = VBox([filter_title, interactive_plot.children[0], q_title, interactive_plot.children[1]], layout=Layout(width='300px', justify_content='center', align_items='flex-start', margin='0 0 0 20px'))

plot_output = interactive_plot.children[-1]

main_layout = HBox([plot_output, controls_box], layout=Layout(align_items='center'))

display(main_layout)
```
